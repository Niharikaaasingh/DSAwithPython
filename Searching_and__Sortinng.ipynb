```python
#BUBBLE SORT ---------------sorting an array

def SortArray(nums):
  n = len(nums)

  for i in range(n):
    isSwap = False
    for j in range(n-i-1):
      if nums[j] > nums[j+1]:
        #swap
        temp = nums[j]
        nums[j] = nums[j+1]
        nums[j+1] = temp
        isSwap = True
    if not isSwap:
      break
  return nums
print(SortArray([5,4,3,2,1]))

```

```python
#INSERTION SORT --------------sorting an array
def SortArray(nums):
  n = len(nums)

  for i in range(1,n):
    key = nums[i]
    j = i-1
    while j>=0 and nums[j]>key:
      nums[j+1] = nums[j]
      j-=1
    nums[j+1]= key
  return nums
print(SortArray([5,4,3,2,1]))

```

```python
#SELECTION SORT -------------sorting an arrray
def SortArray(nums):
  n = len(nums)
  for i in range(n):
    mn = nums[i]
    ind = i
    for j in range(i+1,n):
      if nums[j] < mn:
        mn = nums[j]
        ind = j
    temp = nums[i]
    nums[i] = nums[ind]
    nums[ind] = temp
  return nums
print(SortArray([2,4,6,8,5,9,0]))
```

```python
# QuICK SORT
class Solution:
    def partition(self,nums, l,r):
        key = nums[r]
        start = l
        for i in range(l,r+1):
            if nums[i]<=key:
                temp = nums[i]
                nums[i]=nums[start]
                nums[start]=temp
                start+=1
        return start-1

    def quickSort(self,nums, l,r):
        #base case
        if l>=r:
            return

        p = self.partition(nums,l,r)
        self.quickSort(nums,l,p-1)
        self.quickSort(nums,p+1,r)

    def sortArray(self, nums):
        n = len(nums)
        self.quickSort(nums,0,n-1)
        return nums




```

```python
#LEETCODE 75 ----SORT COLORS
class Solution:
    def sortColors(self, nums):
        """
        Do not return anything, modify nums in-place instead.
        """
        left = 0
        right = len(nums)-1
        i = 0

        while i<=right:
            if nums[i]==1:
                i+=1
            elif nums[i]==0:
                temp = nums[i]
                nums[i]=nums[left]
                nums[left]=temp
                i+=1
                left+=1
            else:
                temp = nums[i]
                nums[i]=nums[right]
                nums[right]=temp
                right-=1


```

```python
#LEETCODE 88 ----MERGE SORTED ARRAY
#You are given two integer arrays nums1 and nums2, sorted in non-decreasing order, and two integers m and n, representing the number of elements in nums1 and nums2 respectively.
#Merge nums1 and nums2 into a single array sorted in non-decreasing order.
#The final sorted array should not be returned by the function, but instead be stored inside the array nums1. To accommodate this, nums1 has a length of m + n, where the first m elements denote the elements that should be merged, and the last n elements are set to 0 and should be ignored. nums2 has a length of n.
class Solution:
    def merge(self, nums1, m, nums2, n):
        """
        Do not return anything, modify nums1 in-place instead.
        """
        i = m-1
        j = n-1
        k = m+n-1

        while j>=0:
            if i<0 or nums2[j]>nums1[i]:
                nums1[k]=nums2[j]
                k-=1
                j-=1
            else:
                nums1[k]=nums1[i]
                k-=1
                i-=1


```

#Leetcode 435 -- Non-overlapping intervals
You are given two integer arrays nums1 and nums2, sorted in non-decreasing order, and two integers m and n, representing the number of elements in nums1 and nums2 respectively.

Merge nums1 and nums2 into a single array sorted in non-decreasing order.

The final sorted array should not be returned by the function, but instead be stored inside the array nums1. To accommodate this, nums1 has a length of m + n, where the first m elements denote the elements that should be merged, and the last n elements are set to 0 and should be ignored. nums2 has a length of n.

```python
class Solution:
    def eraseOverlapIntervals(self, intervals):
        intervals.sort(key = lambda x:x[1])
        n = len(intervals)

        prev = 0
        count = 1

        for i in range(1,n):
            if intervals[i][0]>= intervals[prev][1]:
                count+=1
                prev = i

        return n - count
```
